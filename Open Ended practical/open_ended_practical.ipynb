# Practical 9: Document Vision Application for Scanned Government Documents

## 📌 Objective
A government office requires an automated document vision application to convert scanned physical documents, certificates, identity forms, and official correspondence into editable, structured digital text. 

This notebook implements an end-to-end Computer Vision & OCR pipeline featuring:
1. **Document Image Preprocessing**: Automatic deskewing, noise reduction (Gaussian/Bilateral), adaptive binarization, and perspective correction using OpenCV.
2. **OCR Engine Integration**: Text extraction, confidence scoring, and word bounding box localization with Pytesseract and EasyOCR.
3. **Entity Validation & Regex Parsing**: Automated extraction and validation of Indian Government IDs (Aadhaar, PAN), dates, contact numbers, and key-value fields.
4. **Evaluation Metrics**: Quantitative assessment of OCR accuracy using Character Error Rate (CER) and Word Error Rate (WER).

```python
import sys
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Add core modules to path
sys.path.append(os.path.abspath('.'))
from core.preprocessor import DocumentPreprocessor
from core.ocr_engine import OCREngine
from core.validator import DocumentValidator
from samples.synthetic_docs import create_synthetic_government_doc

%matplotlib inline
print("✓ Core modules successfully imported!")
```

--- 
## 1. Synthetic Document Generation & Dataset Setup
We generate a realistic scanned government verification document with artificial rotation skew (+5.0°), Gaussian noise, and seal stamp overlay.

```python
# Generate synthetic document
raw_doc, ground_truth_text, metadata = create_synthetic_government_doc(skew_angle=5.0, noise_level=12.0)

plt.figure(figsize=(10, 12))
plt.imshow(cv2.cvtColor(raw_doc, cv2.COLOR_BGR2RGB))
plt.title("Input Scanned Government Document (Skewed & Noisy)", fontsize=14)
plt.axis("off")
plt.show()
```

--- 
## 2. Computer Vision Preprocessing Pipeline
The preprocessing pipeline executes key stages:
- **Grayscale Conversion**
- **Automatic Skew Detection & Deskewing** via OpenCV Minimum Bounding Rectangle
- **Denoising** using Gaussian Filtering
- **Binarization** using Otsu's Adaptive Thresholding

```python
preprocessor = DocumentPreprocessor()

preprocess_results = preprocessor.process(
    raw_doc,
    config={
        "auto_deskew": True,
        "denoise_method": "gaussian",
        "denoise_kernel": 5,
        "threshold_method": "otsu"
    }
)

print(f"Detected Document Skew Angle: {preprocess_results['skew_angle']:.2f}°")

# Visualize Pipeline Stages
fig, axes = plt.subplots(2, 2, figsize=(15, 18))
axes[0, 0].imshow(cv2.cvtColor(preprocess_results['original'], cv2.COLOR_BGR2RGB))
axes[0, 0].set_title("1. Original Input Image", fontsize=12)
axes[0, 0].axis("off")

axes[0, 1].imshow(preprocess_results['gray'], cmap='gray')
axes[0, 1].set_title(f"2. Deskewed Grayscale ({preprocess_results['skew_angle']:.2f}°)", fontsize=12)
axes[0, 1].axis("off")

axes[1, 0].imshow(preprocess_results['denoise'], cmap='gray')
axes[1, 0].set_title("3. Denoised Image (Gaussian Filter)", fontsize=12)
axes[1, 0].axis("off")

axes[1, 1].imshow(preprocess_results['binary'], cmap='gray')
axes[1, 1].set_title("4. Binarized Image (Otsu Threshold)", fontsize=12)
axes[1, 1].axis("off")

plt.tight_layout()
plt.show()
```

--- 
## 3. OCR Text Extraction & Bounding Box Localization
We pass the preprocessed image to the OCR Engine to extract text, confidence scores, and bounding box coordinates `[x, y, w, h]`.

```python
ocr = OCREngine()
ocr_output = ocr.extract_text(preprocess_results['final'], engine="auto")

print(f"OCR Engine Used: {ocr_output['engine'].upper()}")
print(f"Total Words Extracted: {ocr_output['total_words']}")
print(f"Average Confidence Score: {ocr_output['avg_confidence'] * 100:.2f}%")

# Visualize Bounding Boxes
bbox_vis = ocr.visualize_boxes(preprocess_results['gray'], ocr_output['words'])

plt.figure(figsize=(12, 14))
plt.imshow(cv2.cvtColor(bbox_vis, cv2.COLOR_BGR2RGB))
plt.title("OCR Bounding Box Localization (Green: High Conf, Yellow: Med, Red: Low)", fontsize=14)
plt.axis("off")
plt.show()
```

--- 
## 4. Government Document Entity Extraction & Validation
We apply regex pattern matchers to extract key entities (Aadhaar, PAN, Dates, Email, Phone) and structured key-value pairs from the OCR text output.

```python
validator = DocumentValidator()
val_results = validator.validate_document(ocr_output['full_text'], ground_truth=ground_truth_text)

print("=== EXTRACTED GOVERNMENT ENTITIES ===")
for entity_type, matches in val_results['entities'].items():
    if matches:
        print(f"• {entity_type.upper()}: {matches}")

print("\n=== STRUCTURED KEY-VALUE PAIRS ===")
for k, v in val_results['key_values'].items():
    print(f"  {k.capitalize()}: {v}")

print("\n=== EVALUATION METRICS ===")
print(f"Character Error Rate (CER): {val_results['cer'] * 100:.2f}%")
print(f"Word Error Rate (WER):      {val_results['wer'] * 100:.2f}%")
```
